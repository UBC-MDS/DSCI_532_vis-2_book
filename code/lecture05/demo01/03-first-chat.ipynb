# Step 3: Your First Chat Conversation

The `chat` object remembers history — follow-up questions use prior context.

```python
from chatlas import ChatGithub
from dotenv import load_dotenv

load_dotenv()

chat = ChatGithub(
    model="gpt-4.1",
    system_prompt="You are a terse assistant.",
)
```

```python
chat.chat("What is the capital of the moon?")
```

```python
# Follow-up — the chat object remembers the previous exchange
chat.chat("Are you sure?")
```

### Inspecting what gets sent

Every API call re-sends the **entire** conversation. Let's see it.

```python
# Full conversation history
for turn in chat.get_turns(include_system_prompt=True):
    print(f"[{turn.role}] {turn}")
```

```python
# Token counts per turn
# Note: tokens_total GROWS because the API resends all prior turns each time
for t in chat.get_tokens():
    print(t)
```

### How big is context, really?

Let's estimate how many tokens a large payload would use.

```python
import json, numpy as np

for n in [10, 100, 1_000, 10_000]:
    big_array = json.dumps(np.random.rand(n).tolist())
    tokens = chat.token_count(big_array)
    print(f"Array of {n:>6} floats → ~{tokens:>6,} tokens  ({len(big_array):>9,} chars)")
```

```python
# For reference: gpt-4.1 has a 1,047,576 token context window
# That 10,000-element array already uses ~10% of it!
tokens_10k = chat.token_count(json.dumps(np.random.rand(10_000).tolist()))
print(f"10k floats: {tokens_10k:,} tokens = {tokens_10k / 1_047_576 * 100:.1f}% of gpt-4.1 context")
```
